# Functions

You have used built-in tools: `print`, `type`, `len`, and `range`. A **function** is a named group of lines you can run whenever you ask. You can make your own.

## Define a function, then call it

To **define** a function means to write the lines and give them a name. The definition stores the lines. It does not run them yet.

`def` starts a definition. Then comes the name, then parentheses, then a colon. The lines inside are indented four spaces, the same way the lines inside `if` are indented.

To **call** a function means to run it. You call it by writing its name with parentheses. Empty parentheses still have to be there. The name alone does not run the function.


```python
def say_hello():
    print("Hello")

say_hello()
```

The parentheses are empty because `say_hello` does not need any extra information. You can call it more than once.

## Parameters and arguments

A **parameter** is a name in the parentheses of the definition. It receives a value when the function is called.

An **argument** is the value you pass in at the call.


```python
def greet(name):
    print(f"Hello, {name}")

greet("Mina")
```

Here `name` is the parameter. `"Mina"` is the argument. Inside the function, `name` holds `"Mina"`.

## return

`print` shows a value in the output. **return** hands a value back to the place that called the function, so you can store it or use it.

When Python reaches `return`, the function ends. Later indented lines inside that function do not run.

If a function reaches its end with no `return`, the call gives back `None`. **None** means "no value." Printing `None` shows the word `None`. The last lesson uses `None` on purpose.


```python
def add(left, right):
    return left + right

total = add(2, 3)
print(total)
```

`left` and `right` are parameters. `2` and `3` are arguments. The call `add(2, 3)` becomes the value `5`, and that value is stored in `total`.

## A function can take a dictionary

A parameter can hold a dictionary. Inside the function you look up a key the same way as outside.


```python
def show_city(person):
    print(person["city"])

show_city({"name": "Sam", "city": "Lagos"})
```
