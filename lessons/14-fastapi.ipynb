# A small contact app

A contact is still a dictionary with the keys `"name"`, `"phone"`, and `"city"`. This lesson builds a small app that returns that list and adds one person. The people stay in a Python list.

Run the cells from top to bottom. Each Try it is one step.


## Route, request, and response

A **route** is a path the app answers. This app uses `/contacts`.

A **request** is what the caller sends to a route.

A **response** is what the route sends back.

`FastAPI()` creates the app.


## Return the list

`@app.get("/contacts")` makes `list_contacts` the route for a request to `/contacts`. The response is the list the function returns.

`TestClient` sends that request from this notebook. `response.json()` is the contact list. The cell finishes. Nothing keeps running after it.


```python
from fastapi import FastAPI
from fastapi.testclient import TestClient

app = FastAPI()
contacts = [
    {"name": "Sam", "phone": "555-0100", "city": "Lagos"},
    {"name": "Mina", "phone": "555-0142", "city": "Seoul"},
]

@app.get("/contacts")
def list_contacts():
    return contacts

with TestClient(app) as client:
    response = client.get("/contacts")
    print(response.json())
```

## Add a person

`@app.post("/contacts")` makes `add_contact` the route that adds one person. The request is the new contact. The response is that person.

`contact` is the input. `: dict` tells you the input should be a dict. It does not change how the function runs. `return contact` is the output.


```python
from fastapi import FastAPI
from fastapi.testclient import TestClient

app = FastAPI()
contacts = [
    {"name": "Sam", "phone": "555-0100", "city": "Lagos"},
    {"name": "Mina", "phone": "555-0142", "city": "Seoul"},
]

@app.get("/contacts")
def list_contacts():
    return contacts

@app.post("/contacts")
def add_contact(contact: dict):
    contacts.append(contact)
    return contact

with TestClient(app) as client:
    response = client.post(
        "/contacts",
        json={"name": "Jo", "phone": "555-0199", "city": "Accra"},
    )
    print(response.json()["name"])
    listed = client.get("/contacts")
    print(len(listed.json()))
```
