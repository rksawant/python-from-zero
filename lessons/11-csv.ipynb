# Saving contacts in a CSV file

A contact is still a dictionary with the keys `"name"`, `"phone"`, and `"city"`. This lesson writes those dictionaries into a file and reads them back. The file remains after the kernel restarts.

Run the cells from top to bottom. Each Try it is one step.


## What a CSV file is

**CSV** is a text file for rows of values. A **row** is one line. Commas separate the values on that line. The file name ends in `.csv`.

The **header** is the first row. It names the columns. Our header is `name,phone,city`.

`csv` is a standard-library module for these files. `import csv` loads it, the same kind of line you used for `pathlib`.


## Write the header

`DictWriter` writes dictionaries as CSV rows. `fieldnames` is the list of keys, in the order they should appear in the header. `writeheader` writes that header row.

`newline=""` tells `open` to let the csv module handle line breaks. Use it when you write a CSV file, and when you create a `DictReader`.


```python
import csv
with open("lessons/contacts.csv", "w", encoding="utf-8", newline="") as csv_file:
    writer = csv.DictWriter(csv_file, fieldnames=["name", "phone", "city"])
    writer.writeheader()
with open("lessons/contacts.csv", "r", encoding="utf-8") as csv_file:
    print(csv_file.read(), end="")
```

## Write one contact

`writerow` writes one dictionary as the next row. The dictionary's keys are the field names.


```python
import csv
contact = {"name": "Sam", "phone": "555-0100", "city": "Lagos"}
with open("lessons/contacts.csv", "w", encoding="utf-8", newline="") as csv_file:
    writer = csv.DictWriter(csv_file, fieldnames=["name", "phone", "city"])
    writer.writeheader()
    writer.writerow(contact)
with open("lessons/contacts.csv", "r", encoding="utf-8") as csv_file:
    print(csv_file.read(), end="")
```

## Write each contact in a list

A `for` loop can call `writerow` once per dictionary.


```python
import csv
contacts = [
    {"name": "Sam", "phone": "555-0100", "city": "Lagos"},
    {"name": "Mina", "phone": "555-0142", "city": "Seoul"},
]
with open("lessons/contacts.csv", "w", encoding="utf-8", newline="") as csv_file:
    writer = csv.DictWriter(csv_file, fieldnames=["name", "phone", "city"])
    writer.writeheader()
    for contact in contacts:
        writer.writerow(contact)
with open("lessons/contacts.csv", "r", encoding="utf-8") as csv_file:
    print(csv_file.read(), end="")
```

## Load the rows back into dictionaries

`DictReader` reads each row as a dictionary. The header supplies the keys. `list(reader)` builds a list with one dictionary per row. Every value comes back as a string, which matches the phone numbers you have been storing.


```python
import csv
with open("lessons/contacts.csv", "r", encoding="utf-8", newline="") as csv_file:
    reader = csv.DictReader(csv_file)
    loaded = list(reader)
print(loaded[0]["name"])
print(len(loaded))
```
