## The one line

A **one-line if/else** picks one value. The value before `if` is used when the test is true. The test sits between `if` and `else`. The value after `else` is used when the test is false.

`label = "in stock" if in_stock else "out of stock"`

`label` is the name that stores the result. `"in stock"` is the value when `in_stock` is true. `in_stock` is the test. `"out of stock"` is the value when the test is false.


```python
in_stock = True
label = "in stock" if in_stock else "out of stock"
print(label)
```
