# Modules

A **module** is a collection of tools kept under one name. `import` loads a module so you can use those tools.

The **standard library** is the set of modules that come with Python. You do not install them with pip. This lesson uses `pathlib` for paths and `datetime` for dates.

Run the cells from top to bottom. Each Try it is one step.


## A path held by Path

`pathlib` is a standard-library module. `Path` is the tool inside it that holds a path. The dot in `pathlib.Path` picks `Path` out of the module.

`as_posix` is a method on a Path. It returns the path with `/` between folder names on every computer.


```python
import pathlib
place = pathlib.Path("lessons/place.txt")
print(place.as_posix())
```

## Write and read with Path

`write_text` writes one string into the file and closes it. `read_text` reads the whole file as one string and closes it. Pass `encoding="utf-8"` to both, the same encoding you used with `open`.


```python
import pathlib
place = pathlib.Path("lessons/place.txt")
place.write_text("Lagos\n", encoding="utf-8")
print(place.read_text(encoding="utf-8"), end="")
```

## Load one tool by name

`from pathlib import Path` loads only `Path`. After that line you write `Path`, not `pathlib.Path`.


```python
from pathlib import Path
note = Path("today-note.txt")
note.write_text("Monday\n", encoding="utf-8")
print(note.read_text(encoding="utf-8"), end="")
```

## Today's date

`datetime` is a standard-library module. `date` is the tool in it for a calendar day. `from datetime import date` loads that tool.

`today` is a method that returns today's date. `year` is the calendar year stored on that date. `str(today)` turns the date into a string, which is the form `write_text` accepts.


```python
from datetime import date
print(date.today())
```

```python
from datetime import date
print(date.today().year)
```
