# Embeddings and a vector database

An **embedding** is a list of numbers that stands for a piece of text. A **vector** is that list. This lesson's embedding counts the letters a through z. The word Notebook lines up with the Notebook sentence more than with the Pen sentence.

A **vector database** stores each text with its vector so you can search later. This lesson uses a list of pairs in memory. That list is a tiny vector database.

**Similarity** is a score for how close two vectors are. The score adds the matching letter counts. A higher score means more of the same letters.

Run the cells from top to bottom. Each Try it is one step. These cells run in the notebook.


```python
def embed(text):
    lowered = text.lower()
    counts = []
    for letter in "abcdefghijklmnopqrstuvwxyz":
        counts.append(lowered.count(letter))
    return counts

def similarity(left, right):
    total = 0
    for a, b in zip(left, right):
        total += a * b
    return total
print(len(embed("Notebook")))
```

```python
# Solution
def embed(text):
    lowered = text.lower()
    counts = []
    for letter in "abcdefghijklmnopqrstuvwxyz":
        counts.append(lowered.count(letter))
    return counts

def similarity(left, right):
    total = 0
    for a, b in zip(left, right):
        total += a * b
    return total
store = []

def add(text):
    store.append({"text": text, "vector": embed(text)})

def search(query):
    if not store:
        return None
    query_vector = embed(query)
    best = None
    best_score = -1
    for row in store:
        score = similarity(query_vector, row["vector"])
        if score > best_score:
            best = row["text"]
            best_score = score
    return best

add("Pen costs 1.5 and is in stock.")
add("Notebook costs 3.0 and is not in stock.")
add("Pencil costs 0.8 and is in stock.")
print(search("notebook"))
```
