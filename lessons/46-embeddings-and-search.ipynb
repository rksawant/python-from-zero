# Embeddings, a vector store, and similarity search

An **embedding** is a list of numbers for a piece of text. A **vector store** keeps each document with its embedding. **Similarity search** embeds the question and returns the stored documents closest to that question.

This lesson uses a small word embedding so the notebook can run with no extra key. The list `WORDS` is the only words that get a number. Each number is `1.0` when that word appears and `0.0` when it does not. `embed_documents` embeds many texts. `embed_query` embeds one question. A hosted embedding model uses those same two methods.

Run the cells from top to bottom. Each Try it is one step.


## The numbers for two notes

The first list of numbers is the Open Pen note. The second list is the Pending Notebook note. The third list is the question `pending`.

Read each list from left to right in the same order as `WORDS`: account, open, pending, pen, notebook, stock, price, vin. The question has `1.0` in the pending slot and `0.0` in the other slots.


```python
from langchain_core.embeddings import Embeddings

WORDS = ("account", "open", "pending", "pen", "notebook", "stock", "price", "vin")

class WordEmbeddings(Embeddings):
    def embed_documents(self, texts):
        return [self.embed_query(text) for text in texts]

    def embed_query(self, text):
        cleaned = text.lower().replace(".", " ").replace(",", " ").replace("?", " ")
        found = cleaned.split()
        return [1.0 if word in found else 0.0 for word in WORDS]

pen = "Account 1001, VIN A100, status Open. Pen price is 1.5 and it is in stock."
book = "Account 1002, VIN B200, status Pending. Notebook price is 3.0 and it is not in stock."
embeddings = WordEmbeddings()
print(WORDS)
print(embeddings.embed_query(pen))
print(embeddings.embed_query(book))
print(embeddings.embed_query("pending"))
```

## Search the vector store

`InMemoryVectorStore` keeps the documents in memory. `from_documents` embeds them and stores them. `similarity_search` returns the closest documents. `k` is how many documents to return.

`similarity_search_with_score` returns each document with a score. A higher score means a closer match.


```python
from langchain_core.documents import Document
from langchain_core.embeddings import Embeddings
from langchain_core.vectorstores import InMemoryVectorStore

WORDS = ("account", "open", "pending", "pen", "notebook", "stock", "price", "vin")

class WordEmbeddings(Embeddings):
    def embed_documents(self, texts):
        return [self.embed_query(text) for text in texts]

    def embed_query(self, text):
        cleaned = text.lower().replace(".", " ").replace(",", " ").replace("?", " ")
        found = cleaned.split()
        return [1.0 if word in found else 0.0 for word in WORDS]

documents = [
    Document(page_content="Account 1001, VIN A100, status Open. Pen price is 1.5 and it is in stock.", metadata={"account": 1001}),
    Document(page_content="Account 1002, VIN B200, status Pending. Notebook price is 3.0 and it is not in stock.", metadata={"account": 1002}),
]
store = InMemoryVectorStore.from_documents(documents, WordEmbeddings())
found = store.similarity_search("pending", k=1)
print(found[0].page_content)
scored = store.similarity_search_with_score("pen", k=1)
print(scored[0][0].page_content)
print(scored[0][1])
```
