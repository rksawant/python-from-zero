# Documents, loaders, and text splitters

A **document** is one piece of writing plus a short record about it. `page_content` is the writing. `metadata` is the record, such as the file name.

A **loader** reads a file and returns documents. A **text splitter** cuts a document into smaller documents. Those smaller documents are **chunks**.

Run the cells from top to bottom. Each Try it is one step.


## Two account documents

Each account note is its own document. The metadata stores the account number. This cell builds the documents in memory. It does not read a file yet.


```python
from langchain_core.documents import Document

documents = [
    Document(page_content="Account 1001, VIN A100, status Open. Pen price is 1.5 and it is in stock.", metadata={"account": 1001}),
    Document(page_content="Account 1002, VIN B200, status Pending. Notebook price is 3.0 and it is not in stock.", metadata={"account": 1002}),
]
for document in documents:
    print(document.metadata["account"])
    print(document.page_content)
    print("---")
```

## Load a file and split it

The cell writes `account_notes.txt`, then loads that file.

The folder comes first. If a folder named `lessons` is in the current folder, the file goes there. That is the case when Jupyter was started from the course folder, the folder that contains `README.md`. If the current folder is already `lessons`, the file goes in the current folder.

`write_text` saves the notes. `load_text` is the loader. It returns one document, and `metadata["source"]` is the file name.

`RecursiveCharacterTextSplitter` cuts that document into chunks. `chunk_size` is the largest number of characters in one chunk. `chunk_overlap` is how many characters one chunk repeats from the chunk before it. Here the overlap is `0`. `split_documents` returns the chunks.


```python
from pathlib import Path
from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter

notes = """Account 1001, VIN A100, status Open. Pen price is 1.5 and it is in stock.
Account 1002, VIN B200, status Pending. Notebook price is 3.0 and it is not in stock.
A recovered account keeps a status. Open means the account is still active. Pending means the account is waiting.
"""
if Path("lessons").is_dir():
    folder = Path("lessons")
else:
    folder = Path(".")
path = folder / "account_notes.txt"
path.write_text(notes, encoding="utf-8")

def load_text(file_path):
    file_path = Path(file_path)
    text = file_path.read_text(encoding="utf-8")
    return [Document(page_content=text, metadata={"source": file_path.name})]

documents = load_text(path)
splitter = RecursiveCharacterTextSplitter(chunk_size=90, chunk_overlap=0)
chunks = splitter.split_documents(documents)
print(path.as_posix())
print(len(chunks))
for chunk in chunks:
    print(chunk.metadata["source"])
    print(chunk.page_content)
    print("---")
```
