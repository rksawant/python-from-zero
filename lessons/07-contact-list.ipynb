## Step 6. Find a person by name

The next teaching cell prints every name. A search does almost the same walk. Instead of printing, it uses `if` and `return`.

If the name matches, `return` that dictionary immediately. `return` ends the function, so the loop stops too. If the loop finishes with no match, `return None`. You met `None` in the functions lesson: it means no value.


```python
contacts = [
    {"name": "Sam", "phone": "555-0100", "city": "Lagos"},
    {"name": "Mina", "phone": "555-0142", "city": "Seoul"},
]
for contact in contacts:
    print(contact["name"])
```

### Try it

Replace `pass` with an `if`. Keep the `if` at the same indent as `pass`. When `contact["name"]` and `name` are the same, return `contact`. Indent that `return` four spaces further than the `if`. Leave the final `return None` where it is.

The cell prints whatever `find_contact` gives back for `"Mina"`. A working function prints the dictionary, not `None`.


```python
def find_contact(contacts, name):
    for contact in contacts:
        if contact["name"] == name:
            return contact        
    return None

contacts = [
    {"name": "Sam", "phone": "555-0100", "city": "Lagos"},
    {"name": "Mina", "phone": "555-0142", "city": "Seoul"},
]
found = find_contact(contacts, "Mina")
print(found)
```

## Step 8. Print every contact

This f-string prints one person. The colon between the name and the phone is a character in the text. It sits inside the quotes. It is a different colon from the one that ends a `for` line.

The f-string itself uses double quotes. Inside it, the dictionary keys use single quotes. `contact['name']` is the same lookup as `contact["name"]`. The single quotes keep the key's quotes from ending the f-string early.


```python
contact = {"name": "Sam", "phone": "555-0100", "city": "Lagos"}
print(f"{contact['name']}: {contact['phone']}")
```
