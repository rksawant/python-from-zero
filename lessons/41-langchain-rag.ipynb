# A LangChain RAG path

**LangChain** is a library that connects the steps from a document to an answer.

The path in this lesson is:

1. Turn the document into text.
2. Chunk that text.
3. Build an embedding for each chunk.
4. Store the chunks in a vector store.
5. Run a similarity search for the user's query.
6. Build one prompt from the instructions, the retrieved text, and the query.
7. Send that prompt to the model when `GROQ_API_KEY` is set.

The embedding is the letter count from lesson 40. The **vector store** is the vector database from lesson 40, kept in memory by LangChain.

## Vector store and similarity search

**Similarity search** finds the stored chunk closest to the query. `k` is how many chunks to return. This step asks for 1.


### Try it

Replace `"pencil"` with `"notebook"`. Before the change the cell prints `Pencil costs 0.8 and is in stock.` After the change it prints `Notebook costs 3.0 and is not in stock.`


```python
from langchain_core.embeddings import Embeddings
from langchain_core.vectorstores import InMemoryVectorStore
from langchain_text_splitters import RecursiveCharacterTextSplitter

class LetterEmbeddings(Embeddings):
    def embed_documents(self, texts):
        return [self.embed_query(text) for text in texts]

    def embed_query(self, text):
        lowered = text.lower()
        return [lowered.count(letter) for letter in "abcdefghijklmnopqrstuvwxyz"]

document = """Pen costs 1.5 and is in stock.
Notebook costs 3.0 and is not in stock.
Pencil costs 0.8 and is in stock.
"""
splitter = RecursiveCharacterTextSplitter(chunk_size=40, chunk_overlap=10)
chunks = splitter.split_text(document)
store = InMemoryVectorStore(LetterEmbeddings())
store.add_texts(chunks)
found = store.similarity_search("pencil", k=1)
print(found[0].page_content)
```

```python
# Solution
from langchain_core.embeddings import Embeddings
from langchain_core.vectorstores import InMemoryVectorStore
from langchain_text_splitters import RecursiveCharacterTextSplitter

class LetterEmbeddings(Embeddings):
    def embed_documents(self, texts):
        return [self.embed_query(text) for text in texts]

    def embed_query(self, text):
        lowered = text.lower()
        return [lowered.count(letter) for letter in "abcdefghijklmnopqrstuvwxyz"]

document = """Pen costs 1.5 and is in stock.
Notebook costs 3.0 and is not in stock.
Pencil costs 0.8 and is in stock.
"""
splitter = RecursiveCharacterTextSplitter(chunk_size=40, chunk_overlap=10)
chunks = splitter.split_text(document)
store = InMemoryVectorStore(LetterEmbeddings())
store.add_texts(chunks)
found = store.similarity_search("notebook", k=1)
print(found[0].page_content)
```

## Send the prompt

This step uses Groq, the service from lesson 38. The system message is the instructions. The user message is the retrieved text plus the query. When `GROQ_API_KEY` is missing, the cell prints a message and does not send a request.


```python
import os
from groq import Groq

instructions = "Answer in one sentence. Use only the retrieved text."
retrieved = "Notebook costs 3.0 and is not in stock."
query = "notebook"
key = os.environ.get("GROQ_API_KEY")
if not key:
    print("GROQ_API_KEY is missing. Set it in the environment, then run this cell again.")
else:
    client = Groq(api_key=key)
    response = client.chat.completions.create(
        model="openai/gpt-oss-120b",
        messages=[
            {"role": "system", "content": instructions},
            {
                "role": "user",
                "content": "Retrieved text: " + retrieved + "\nQuestion: " + query,
            },
        ],
    )
    print(response.choices[0].message.content)
```
