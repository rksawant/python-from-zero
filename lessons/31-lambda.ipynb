# lambda

A **lambda** is a function written on one line. The name before the colon is the input. The expression after the colon is the value it returns.

`with_tax = lambda price: price + 0.1`

`with_tax` is the name of the function. `price` is the input. `price + 0.1` is the value that comes back. A Pen at 1.5 becomes 1.6.

Run the cells from top to bottom. Each Try it is one step. These cells run in the notebook.


```python
with_tax = lambda price: price + 0.1
print(with_tax(1.5))

sqr = lambda inp: inp*inp  if type(inp) == int else "Enter a valid integer"
print(sqr(-2.1))
```
