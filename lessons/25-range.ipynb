# range

**range** is a sequence of whole numbers. It stores the start, the stop, and the step. It builds the full list when you ask for one.

`range(stop)` starts at 0 and stops before `stop`. The step is 1. `range(3)` is 0, 1, 2.

`range(start, stop)` starts at `start` and stops before `stop`. The step is 1. `range(1, 4)` is 1, 2, 3.

`range(start, stop, step)` uses all three parts. `step` is how much to add each time. `range(0, 6, 2)` is 0, 2, 4.

`list` around a range builds a list of those numbers. A `for` loop walks the numbers one at a time. Lesson 5 introduced `for`. Here the loop walks a range.
