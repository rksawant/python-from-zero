# Prompts and a Groq chat model

A **prompt** is the text you send to the model. A **chat model** is a model that reads a list of messages and writes a reply. **LangChain** is the library that builds those messages and calls the model. **ChatGroq** is the LangChain chat model that calls Groq.

The model name is `openai/gpt-oss-120b`, the same model as lesson 38.

`GROQ_API_KEY` is the name of your key. The key stays in the environment, outside this file.

Open PowerShell. Any folder works. This command does not need `(.venv)`.

```powershell
$env:GROQ_API_KEY = "paste-your-key-here"
```

`$env:` means an environment variable in this window. `GROQ_API_KEY` is the name. The text after `=` is your key. The setting lasts until you close the window. Start Jupyter from that same window if you want this notebook to see the key.

Run the cells from top to bottom. Each Try it is one step.


## Two messages

A **system message** tells the model how to answer. A **human message** is the question.

`SystemMessage` holds the system message. `HumanMessage` holds the human message. `model.invoke` sends the list. `reply.content` is the text that comes back.

When `GROQ_API_KEY` is missing, the cell prints a message and does not send a request.


````python
import os
from langchain_core.messages import HumanMessage, SystemMessage
from langchain_groq import ChatGroq

key = os.environ.get("GROQ_API_KEY")
if not key:
    print("GROQ_API_KEY is missing. Set it in the environment, then run this cell again.")
else:
    model = ChatGroq(model="openai/gpt-oss-120b", api_key=key, temperature=0)
    messages = [
        SystemMessage(content="You answer in one sentence."),
        HumanMessage(content="Say hello in one sentence."),
    ]
    reply = model.invoke(messages)
    print(reply.content)
````

## A prompt about an account note

The note is one sentence about a recovered account. The human message asks for the price of the Pen. The system message tells the model to use the note and to answer in one sentence.


````python
import os
from langchain_core.messages import HumanMessage, SystemMessage
from langchain_groq import ChatGroq

note = "Account 1001, VIN A100, status Open. Pen price is 1.5 and it is in stock."
key = os.environ.get("GROQ_API_KEY")
if not key:
    print("GROQ_API_KEY is missing. Set it in the environment, then run this cell again.")
else:
    model = ChatGroq(model="openai/gpt-oss-120b", api_key=key, temperature=0)
    messages = [
        SystemMessage(content="Answer in one sentence using only the note."),
        HumanMessage(content="Note: " + note + " Question: What is the price of the Pen?"),
    ]
    reply = model.invoke(messages)
    print(reply.content)
````
