A **class** is the block from lesson 16. This lesson uses a class to share functions.

**Inheritance** means one class receives the functions of another class. The class in the parentheses is the **parent**. The new class is the **child**.

## Multiple inheritance

**Multiple inheritance** means one child lists more than one parent. `class StockedProduct(Product, Priced)` receives `label` from Product and `price_line` from Priced.


```python
class Product:
    def label(self):
        return "product"

class Priced:
    def price_line(self):
        return "1.5"

class StockedProduct(Product, Priced):
    pass

item = StockedProduct()
print(item.label())
print(item.price_line())
```

## Multilevel inheritance

**Multilevel inheritance** means the parent has a parent of its own. Named is the first parent. Item is a child of Named. Pen is a child of Item. Pen receives `kind` through Item.


### Try it

Replace `class Pen(Named)` with `class Pen(Item)`. Before the change the cell prints `named`. After the change it prints `item`.


```python
class Named:
    def kind(self):
        return "named"

class Item(Named):
    def kind(self):
        return "item"

class Pen(Named):
    pass

print(Pen().kind())
```

### Solution

Try the exercise above first. Unfold the next cell the same way as in lesson 1: click the blue bar on its left, then press Shift+Enter.


```python
# Solution
class Named:
    def kind(self):
        return "named"

class Item(Named):
    def kind(self):
        return "item"

class Pen(Item):
    pass

print(Pen().kind())
```
