# Sets

A **set** is a group of values with no repeats. Curly braces with values build one: `{"Pen", "Notebook"}`. `{}` is an empty dict. An empty set is `set()`.

`sorted` puts names in alphabetical order. A set does not promise an order, so a cell that prints `sorted(...)` stays still when you run it again.

Run the cells from top to bottom. Each Try it is one step. These cells run in the notebook.


```python
desk = {"Pen", "Notebook"}
print("Pen" in desk)
print(len(desk))
```

## union

`union` returns a new set with every value that is in either set. `first` holds Pen and Notebook. `second` holds Notebook and Pencil.

```python
first = {"Pen", "Notebook"}
second = {"Notebook", "Pencil"}
both = set()
both = first.union(second)
print(sorted(both))
```
