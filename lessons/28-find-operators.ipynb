# Other find operators

Lessons 20, 21, and 27 practiced a set of find operators. This lesson lists the rest in groups. Four of them get a tiny example you can run: `$exists`, `$type`, `$mod`, and `$not`.

Pen has qty 10. Notebook has qty 3 and `color` set to `blue`. Pencil has qty 4. Only Notebook has a color.

Run the cells from top to bottom. Each Try it is one step.


## Comparison, lesson 20

- `$eq` means equal.
- `$ne` means not equal.
- `$gt` means greater than.
- `$gte` means greater than or equal.
- `$lt` means less than.
- `$lte` means less than or equal.
- `$in` means the value is in the list.
- `$nin` means the value is outside the list.


## Logical, lesson 21

- `$and` means every filter in the list matches.
- `$or` means one filter in the list matches.
- `$nor` means none of the filters match.
- `$not` means the comparison inside it fails.


## Element

`$exists` checks whether the field is present. `{"color": {"$exists": True}}` keeps documents that have a color.

`$type` checks the kind of value. `{"qty": {"$type": "int"}}` keeps a qty stored as a whole number. `{"qty": {"$type": "string"}}` looks for text, so these quantities do not match.

## Evaluation

`$mod` divides a number and checks the remainder. `{"qty": {"$mod": [2, 0]}}` keeps even quantities. The list is divisor, then remainder. Pen at 10 and Pencil at 4 match. Notebook at 3 does not.

`$regex` matches a text pattern. Lesson 27 uses it on the product name.

- `$expr` compares values on the same document, for example `{"$expr": {"$gt": ["$price", 1]}}`.
- `$jsonSchema` checks the document against a JSON Schema.
- `$text` searches a text index.
- `$where` runs a JavaScript test on the document.


### Try it

Replace `[2, 1]` with `[2, 0]`. Before the change the remainder is 1, so the cell prints `Notebook`. After the change the remainder is 0, so the cell prints `Pen` and `Pencil`.


```python
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017", serverSelectionTimeoutMS=2000)
products = client["python_course"]["products"]
products.delete_many({})
products.insert_many([
    {"name": "Pen", "price": 1.5, "in_stock": True, "qty": 10},
    {"name": "Notebook", "price": 3.0, "in_stock": False, "qty": 3, "color": "blue"},
    {"name": "Pencil", "price": 0.8, "in_stock": True, "qty": 4},
])
for doc in products.find({"qty": {"$mod": [2, 1]}}):
    print(doc["name"])
```

## Array, lesson 27

- `$size` matches a list with this many items.
- `$all` matches a list that contains every listed value.
- `$elemMatch` matches when one list item meets every condition inside it.


## Geospatial

These operators need a location field. This course does not store one.

- `$geoWithin` means the point is inside a shape.
- `$geoIntersects` means the geometry overlaps a shape.
- `$near` means close to a point, nearest first.
- `$nearSphere` means close to a point, measured on a sphere.


## Bitwise

These operators test bits in an integer. A bit is one binary digit, 0 or 1.

- `$bitsAllClear` means every listed bit is 0.
- `$bitsAllSet` means every listed bit is 1.
- `$bitsAnyClear` means at least one listed bit is 0.
- `$bitsAnySet` means at least one listed bit is 1.
