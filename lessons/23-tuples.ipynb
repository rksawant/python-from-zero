# Tuples

A **tuple** is an ordered group of values that cannot be changed. Parentheses build one: `("Pen", 1.5, True)`. The first position is `0`, then `1`, then `2`.

Run the cells from top to bottom. Each Try it is one step. These cells run in the notebook.


## Packing

**Packing** puts values into one tuple. `item = ("Pen", 1.5, True)` packs the name, the price, and `in_stock`.


```python
item = ("Pen", 1.5, True)
print(item[0])
print(item[1])
print(item[2])
```

## Unpacking

**Unpacking** takes the values back out. `name, price, in_stock = item` stores each position in its own name. The count of names on the left has to match the count of values in the tuple.


### Try it

Replace the three lines that set `None` with `name, price, in_stock = item`. Before the change the cell prints `None None None`. After the change it prints `Pen 1.5 True`.


```python
item = ("Pen", 1.5, True)
name = None
price = None
in_stock = None
print(name, price, in_stock)
```

### Solution

Try the exercise above first. Unfold the next cell the same way as in lesson 1: click the blue bar on its left, then press Shift+Enter.


```python
# Solution
item = ("Pen", 1.5, True)
name, price, in_stock = item
print(name, price, in_stock)
```

## Why a tuple cannot be changed

Assigning a new value to a position asks Python to change the tuple. Python raises **TypeError** for that assignment. The original value stays in place.


### Try it

Replace `kind = None` with `kind = type(error).__name__`. Before the change the cell prints `None` and `Pen`. After the change it prints `TypeError` and `Pen`.


```python
item = ("Pen", 1.5, True)
try:
    item[0] = "Pencil"
except TypeError as error:
    kind = None
print(kind)
print(item[0])
```

### Solution

Try the exercise above first. Unfold the next cell the same way as in lesson 1: click the blue bar on its left, then press Shift+Enter.


```python
# Solution
item = ("Pen", 1.5, True)
try:
    item[0] = "Pencil"
except TypeError as error:
    kind = type(error).__name__
print(kind)
print(item[0])
```
