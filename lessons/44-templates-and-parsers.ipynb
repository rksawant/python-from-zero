# Prompt templates and output parsers

A **prompt template** is a message with blanks. You fill the blanks, then send the messages. An **output parser** turns the model's text into a value, such as a string or a dictionary.

`{status}` is one blank. The name inside the braces is the blank's name.

Run the cells from top to bottom. Each Try it is one step.


## Fill the blanks

`ChatPromptTemplate.from_messages` builds the template. `format_messages` fills the blanks and returns the messages. `message.type` is `system` or `human`. `message.content` is the filled text.

This cell does not call Groq. It only fills the template.


```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages([
    ("system", "You answer in one sentence."),
    ("human", "Status is {status}. Item is {name}. Price is {price}. In stock is {in_stock}."),
])
messages = prompt.format_messages(status="Open", name="Pen", price=1.5, in_stock=True)
for message in messages:
    print(message.type + ": " + message.content)
```

## Parse a JSON string

**JSON** is text that stores names and values. `JsonOutputParser` reads a JSON string and returns a dictionary.

This cell does not call Groq. The string is already in the cell.


```python
from langchain_core.output_parsers import JsonOutputParser

parser = JsonOutputParser()
text = '{"name": "Notebook", "price": 3.0, "in_stock": false}'
print(parser.parse(text))
```

## Send the template to Groq

`StrOutputParser` returns the model's reply as one string. The `|` symbol connects the steps. `prompt | model | parser` fills the template, calls Groq, then returns the string.

The template asks for the price in the account note. When `GROQ_API_KEY` is missing, the cell prints a message and does not send a request. Lesson 43 shows how to set the key in PowerShell.


```python
import os
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate
from langchain_groq import ChatGroq

prompt = ChatPromptTemplate.from_messages([
    ("system", "You answer in one sentence."),
    ("human", "Status is {status}. Item is {name}. Price is {price}. In stock is {in_stock}. What is the price?"),
])
key = os.environ.get("GROQ_API_KEY")
if not key:
    print("GROQ_API_KEY is missing. Set it in the environment, then run this cell again.")
else:
    model = ChatGroq(model="openai/gpt-oss-120b", api_key=key, temperature=0)
    chain = prompt | model | StrOutputParser()
    answer = chain.invoke({
        "status": "Pending",
        "name": "Notebook",
        "price": 3.0,
        "in_stock": False,
    })
    print(answer)
```
