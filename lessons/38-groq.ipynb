# A small Groq program

**GenAI** is a program that asks a model to write text. **Groq** is a service that runs the model and returns that text.

An **environment variable** is a name stored by the system, outside the file. `GROQ_API_KEY` is the name of your key. `os.environ.get("GROQ_API_KEY")` reads it. The key stays out of the lesson file.

Open PowerShell. Any folder works. This command does not need `(.venv)`.

```powershell
$env:GROQ_API_KEY = "paste-your-key-here"
```

`$env:` means an environment variable in this window. `GROQ_API_KEY` is the name. The text after `=` is your key. The setting lasts until you close the window. Start Jupyter from that same window if you want the notebook to see the key.

Run the cells from top to bottom. Each Try it is one step.


## The call

`Groq(api_key=key)` opens the client. `api_key` is the key from the environment.

`client.chat.completions.create` sends the request. `model` is the name of the model Groq should run. `messages` is the list of texts. The system message is the instruction. The user message is the question.

`response.choices[0].message.content` is the text that comes back.

When `GROQ_API_KEY` is missing, the cell prints a message and does not send a request.


````python
import os
from groq import Groq

key = os.environ.get("GROQ_API_KEY")
if not key:
    print("GROQ_API_KEY is missing. Set it in the environment, then run this cell again.")
else:
    client = Groq(api_key=key)
    response = client.chat.completions.create(
        model="openai/gpt-oss-120b",
        messages=[
            {"role": "system", "content": "You answer in one sentence."},
            {"role": "user", "content": "Say hello in one sentence."},
        ],
    )
    print(response.choices[0].message.content)
````
