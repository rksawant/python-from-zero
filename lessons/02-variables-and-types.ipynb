## A value

A **value** is one piece of information. These are values:

- `42`
- `3.5`
- `"Sam"`

The quotes around `Sam` mark text, the same way they did for `print` in the first notebook.


## A variable

A **variable** is a name that holds a value. You choose the name so you can use the value again.

A useful name describes the value. `age` is a useful name for a person's age. `x` is allowed, and it tells you less.

A name uses letters. `age` and `Age` are two different names, because one starts with a small `a` and the other with a capital `A`. If you want two words in a name, join them with `_`, the **underscore**. `favorite_color` is one name. A space is not allowed in a name.


## Assignment

**Assignment** stores a value under a name. You write the name, then a single `=`, then the value.

```python
age = 34
```

Read that line as: the name `age` now holds `34`.

A single `=` stores the value on the right into the name on the left. In arithmetic, people also use `=` to ask whether two sides are the same. Python uses two equals signs, `==`, for that question. This lesson uses one equals sign to store a value.

To see the value a name holds, put the name inside `print`'s parentheses, with no quotes around the name. Quotes would show the letters of the name itself.


````python
age = 34
print(age)
print("age")
````

The first output line is the value `34`. The second line is the word `age`, because of the quotes.

The next cell does not store `age` again. It uses the `age` remembered from the cell above. Run that cell first, then run this one. If the output is red and the last line says `NameError`, the kernel does not know the name yet. A **NameError** means you used a name you have not stored in a cell the kernel has run.


````python
print(age)
````

## Types

A **type** is the kind of value. Python keeps track of the kind. The tool `type` reports it. You put a value, or a name that holds a value, inside the parentheses, then print the report.

### Integers

An **integer** is a whole number, with no decimal point. It can be negative. `34`, `0`, and `-2` are integers. Python's name for this type is `int`.

For an integer, `type` prints a line like `<class 'int'>`. The word you need is `int`. The word `class` is part of how Python labels the report. You can set the word `class` aside in this course.


````python
print(type(34))
````

The output above is the type report for `34`.

## Store a new value in the same name

Assignment can happen again with a name you already used. The name then holds the new value.


````python
city = "Dakar"
print(city)
city = "Lima"
print(city)
````

After the second assignment, `city` holds `Lima`. The printed lines show the value before the change, then the value after it.

## Arithmetic

You can calculate with numbers. `+` adds. `-` subtracts. `*` multiplies. `/` divides.

An **expression** is a calculation written with values and symbols. Python works out the result. You can print an expression, and you can store it in a variable.


````python
print(2 + 3)
print(10 - 4)
print(2 * 3)
price = 4
tax = 1
print(price + tax)
````

Division produces a float, including when the numbers divide evenly.


````python
print(6 / 2)
````
