## try and except

An **exception** is an error Python can hand to your code. The last line of the red text starts with the exception's name.

`try` runs its indented lines and watches for an exception. `except` names one exception, then a colon. Its indented lines run only when that exception happens. When the `try` lines finish with no exception, Python skips the `except` lines.


## A missing file

`FileNotFoundError` is the exception for a path that does not point at a file.


```python
try:
    with open("lessons/missing-note.txt", "r", encoding="utf-8") as note:
        print(note.read(), end="")
except FileNotFoundError:
    print("That file is not there yet.")
```

## A value with the wrong content

`int` can turn a string of digits into an integer. `int("12")` is the integer `12`. A newline after the digits is allowed. `int("lots")` is not allowed.

`ValueError` is the exception when a value has the wrong content for the tool you asked to use.


```python
print(int("12"))
```

```python
try:
    print(int("12"))
except ValueError:
    print("That is not a whole number.")
```

The `except` lines did not run. `"12"` is a whole number, so `try` finished.


```python
try:
    print(int("lots"))
except ValueError:
    print("That is not a whole number.")
```
