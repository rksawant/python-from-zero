# Pinecone

**Pinecone** is a service that stores vectors and searches them. An **index** is the named store. This lesson uses the index `python-course`. The **dimension** is the length of each vector. The letter-count embedding from lesson 40 has length 26. The **metric** is how closeness is scored. This lesson uses cosine.

**Upsert** writes a record. The record has an `id`, a `values` vector, and `metadata` such as the product name and price.

**Similarity search** sends a vector and reads the closest records. `top_k` is how many records to return.

The key stays out of the file. `PINECONE_API_KEY` is the environment variable that holds it.

Open PowerShell. Any folder works. This command does not need `(.venv)`.

```powershell
$env:PINECONE_API_KEY = "paste-your-key-here"
```

`$env:` means an environment variable in this window. `PINECONE_API_KEY` is the name. The text after `=` is your key. The setting lasts until you close the window. Start Jupyter from that same window if you want the notebook to see the key.

When the key is missing, the live cell prints a message and does not call Pinecone. The first time the key is set, that cell creates the index and waits until it is **ready**, which means Pinecone can take a write.

## The upsert record

`id` is the name of the record. `values` is the vector. `metadata` holds the product fields you want back from a search.


````python
def embed(text):
    lowered = text.lower()
    counts = []
    for letter in "abcdefghijklmnopqrstuvwxyz":
        counts.append(float(lowered.count(letter)))
    return counts
record = {
    "id": "notebook",
    "values": embed("notebook"),
    "metadata": {"name": "Notebook", "price": 3.0},
}
print(record["id"])
print(len(record["values"]))
print(record["metadata"]["name"])
print(record["metadata"]["price"])
````

## Index, upsert, and search


````python
import os
import time
from pinecone import Pinecone, ServerlessSpec

def embed(text):
    lowered = text.lower()
    counts = []
    for letter in "abcdefghijklmnopqrstuvwxyz":
        counts.append(float(lowered.count(letter)))
    return counts

key = os.environ.get("PINECONE_API_KEY")
if not key:
    print("PINECONE_API_KEY is missing. Set it in the environment, then run this cell again.")
else:
    pc = Pinecone(api_key=key)
    name = "python-course"
    if not pc.has_index(name):
        pc.create_index(
            name=name,
            dimension=26,
            metric="cosine",
            spec=ServerlessSpec(cloud="aws", region="us-east-1"),
        )
        ready = False
        while not ready:
            info = pc.describe_index(name)
            ready = info.status.ready
            if not ready:
                time.sleep(2)
    index = pc.Index(name)
    vector = embed("notebook")
    index.upsert(vectors=[{
        "id": "notebook",
        "values": vector,
        "metadata": {"name": "Notebook", "price": 3.0},
    }])
    found = index.query(vector=vector, top_k=1, include_metadata=True)
    print(found.matches[0].metadata["name"])
````
