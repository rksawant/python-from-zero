So far one variable has held one value. This lesson stores several values in a **list**, and stores labeled pairs in a **dictionary**. People often shorten dictionary to **dict**. This lesson uses the full word.

Run the cells from top to bottom. Positions start at `0`, the same rule as for characters in a string.


## A list

A **list** keeps values in order. You write the values inside square brackets, with a comma between values.


```python
colors = ["red", "blue", "green"]
print(colors)
```

Each value in the list is an **item**. `"red"` is the first item.

When Python prints a list, it shows single quotes around each string. Those values are still strings. Keep using double quotes when you type your own cells.

`len` works on a list. It tells you how many items are in the list.

You read one item the same way you read one character in a string. Write the list name, then square brackets, with the position inside. Position `0` is the first item. Position `2` is the third item.


```python
colors = ["red", "blue", "green"]
print(len(colors))
print(colors[0])
print(colors[2])
```

## Add an item to the end

`.append()` is a method on a list, the same kind of attached tool as `.upper()` on a string. It adds one item at the end and changes that list.


```python
colors = ["red", "blue"]
colors.append("green")
print(colors)
```

## A dictionary

A **dictionary** stores pairs. Each pair has a **key** and a **value**. The key is the label you look up. The value is what is stored under that label. You already used the word value for any piece of information. Here it also means the stored side of one pair.

You write a dictionary inside curly braces. Each pair is written `key: value`. A comma separates pairs.


```python
book = {"title": "Pachinko", "pages": 480}
print(book["title"])
print(book["pages"])
```

The lookup uses square brackets, like a list position, except you write the key instead of a position number. The key has to match exactly, including capital letters.

If the key is missing, the last line of the red error says `KeyError`. A **KeyError** means that label is not in the dictionary. Check the spelling.

## Add a pair

To add a pair, assign to a new key with a single `=`.


```python
book = {"title": "Pachinko"}
book["pages"] = 480
print(book["pages"])
```

## A list of dictionaries

A list item can be a dictionary. The first item is still position `0`.

Square brackets can span several lines. Extra spaces and line breaks between items do not change the list. The commas still separate the items. A comma after the last item is allowed, and it does not add an empty item. The spaces at the start of the inner lines are for reading. Python does not require them inside the brackets.


```python
people = [
    {"name": "Sam", "city": "Lagos"},
    {"name": "Mina", "city": "Seoul"},
]
print(people[0]["name"])
```

Read `people[0]["name"]` from left to right. `people[0]` is the first dictionary. `["name"]` looks up the name inside that dictionary.
