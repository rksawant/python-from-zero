# Insert and find products

A product document has `name`, `price`, and `in_stock`. This lesson stores those documents in the `products` collection and reads them back.

Run the cells from top to bottom. Each Try it is one step.


## insert_one and find_one

**insert_one** stores one document. **find_one** returns one document. The dict you pass to `find_one` is the filter. `{"name": "Pen"}` keeps the document whose name is Pen.

**delete_many** removes every document that matches its filter. `{}` matches every document. The cell clears `products` first so a second run does not store a second Pen.

The key `_id` is the id MongoDB adds when you insert a document. An **ObjectId** is the kind of value MongoDB puts there. A new run prints a new ObjectId.


```python
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017", serverSelectionTimeoutMS=2000)
products = client["python_course"]["products"]
products.delete_many({})
products.insert_one({"name": "Pen", "price": 1.5, "in_stock": True})
doc = products.find_one({"name": "Pen"})
print(doc["name"])
print(doc["price"])
print(doc["in_stock"])
print(doc["_id"])
```
