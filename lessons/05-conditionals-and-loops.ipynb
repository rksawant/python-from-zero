# Conditionals and loops

A **conditional** is a choice: Python runs one group of lines, or a different group, based on a question. A **loop** repeats lines. This lesson teaches both, one small step at a time.

Run the cells from top to bottom.


## A comparison

A **comparison** asks a question about two values. The answer is a boolean, `True` or `False`, which you used in the variables lesson.

- `==` asks "are these the same?"
- `!=` asks "are these different?"
- `>` asks "is the left one greater?"
- `<` asks "is the left one smaller?"
- `>=` asks "is the left one greater, or are they the same?"
- `<=` asks "is the left one smaller, or are they the same?"

One `=` stores a value. Two equals signs `==` ask a question.


```python
print(5 > 3)
print(5 == 3)
print(5 != 3)
print("cat" == "cat")
```

## if

A **condition** is a comparison, or another expression, that comes out as `True` or `False`.

`if` runs the following indented lines only when the condition is `True`.

**Indent** means spaces at the start of a line. Python uses the indent to see which lines belong inside the `if`. Use four spaces. In a code cell, press the Tab key and Jupyter inserts those spaces.

The `if` line ends with a colon `:`. The colon means the indented lines that follow are the inside.


```python
temperature = 30
if temperature > 25:
    print("warm")
```

If `temperature` held `20`, this cell would print nothing. The condition would be `False`, and Python would skip the indented `print`.

## else

`else` runs when the `if` condition is `False`. The word `else` lines up with `if`, with no indent. `else` ends with a colon. Its own lines are indented.


```python
temperature = 20
if temperature > 25:
    print("warm")
else:
    print("cool")
```

## elif

`elif` is short for "else if". It is another condition, checked only when the `if` above it was `False`. Put it between `if` and `else`. Python checks from top to bottom and runs the first matching choice.


```python
temperature = 10
if temperature > 25:
    print("warm")
elif temperature > 15:
    print("mild")
else:
    print("cool")
```

Here `10 > 25` is `False`, and `10 > 15` is `False`, so Python prints `cool`.

## for

A `for` **loop** walks through a list, one item at a time. Each time around, it stores the current item in a name you choose. The lines inside the loop are indented, and the `for` line ends with a colon.


```python
colors = ["red", "blue", "green"]
for color in colors:
    print(color)
```

Read the first line as: for each item in `colors`, give that item the name `color`, then run the indented lines. `color` is a normal variable. It holds one item at a time.

## range

`range` is a built-in tool that produces integers. `range(3)` produces `0`, `1`, and `2`. It stops before the number you give it, the same way a slice stops before its end position.

`range(1, 4)` produces `1`, `2`, and `3`. The first number is the start. The second number is the stop, and the stop is not included.


```python
for number in range(1, 4):
    print(number)
```

## while

A `while` loop repeats as long as its condition is `True`. Something inside the loop has to change so the condition can become `False`. Otherwise the kernel stays busy and the label shows `In [*]:`.

If that happens, open the **Kernel** menu and choose **Interrupt Kernel**. If you then need a fresh start, choose **Restart Kernel** and run the notebook from the top.


```python
count = 1
while count < 4:
    print(count)
    count = count + 1
```

Each time around, `count` goes up by 1. When `count` becomes `4`, the comparison `count < 4` is `False`, and the loop stops.


### Try it

```python
# Your turn.
count = 1
while count < 3:
    print(count)
    count = count + 1
    
```
