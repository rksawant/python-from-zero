# Routes in two files

A contact is still a dictionary with the keys `"name"`, `"phone"`, and `"city"`. A product is a dictionary with the keys `"name"`, `"price"`, and `"in_stock"`. `in_stock` is True when the shop has the product, and False when it does not.

The routes live in `routes.py`. `contacts_app.py` attaches those routes to the app.

## A router

A **router** is a group of routes. **APIRouter** creates a router. `APIRouter()` makes an empty one.

`routes.py` already has a route named `/all_contacts`. That route returns a dictionary with the key `"message"`.


```python
from fastapi import APIRouter, FastAPI
from fastapi.testclient import TestClient

contacts = [
    {"name": "Sam", "phone": "555-0100", "city": "Lagos"},
    {"name": "Mina", "phone": "555-0142", "city": "Seoul"},
]

router = APIRouter()

@router.get("/all_contacts")
def get_all_contacts():
    return {"message": "All contacts"}

@router.get("/first_contact")
def first_contact():
    return contacts[0]

app = FastAPI()
app.include_router(router)

with TestClient(app) as client:
    response = client.get("/all_contacts")
    print(response.json())

    response = client.get("/first_contact")
    print(response.json())
```

## Put the routers on the app

**include_router** puts a router's routes on the app. The call is `app.include_router(router)`.

The **status code** is the number on the response. `200` means the route ran. `404` means the app has no route at that path.

`contacts_app.py` calls it twice: once for `router`, once for `product_router`.


```python
from fastapi import APIRouter, FastAPI
from fastapi.testclient import TestClient

router = APIRouter()

@router.get("/all_contacts")
def get_all_contacts():
    return {"message": "All contacts"}

product_router = APIRouter()
products = [
    {"name": "Pen", "price": 1.5, "in_stock": True},
    {"name": "Notebook", "price": 3.0, "in_stock": False},
]

@product_router.get("/products")
def list_products():
    return products

app = FastAPI()
app.include_router(router)
app.include_router(product_router)

with TestClient(app) as client:
    print(client.get("/all_contacts").json())
    print(client.get("/products").status_code)
```

## The uvicorn command stays

`uvicorn` loads the file `contacts_app.py` and the name `app`. `include_router` has already put the routes from `routes.py` onto that app. The command stays `uvicorn contacts_app:app`.

Run it from the `lessons` folder, while the start of the line shows `(.venv)`. The `lessons` folder holds `contacts_app.py` and `routes.py`. The venv holds `uvicorn`. The PowerShell window stays busy. In the browser, open `http://127.0.0.1:8000/docs`. Ctrl+C stops the app.
