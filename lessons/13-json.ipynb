# Saving contacts as JSON

A contact is still a dictionary with the keys `"name"`, `"phone"`, and `"city"`. This lesson writes that list into a file and reads it back. The file remains after the kernel restarts.

Run the cells from top to bottom. Each Try it is one step.


## JSON is text

**JSON** is data written as text. A JSON file is that text saved in a file. The name ends in `.json`.

`json` is a standard-library module for this text. `import json` loads it, the same kind of line you used for `csv`.


## Turn the list into text

`dumps` turns a Python value into JSON text. The call is `json.dumps`.


```python
import json
contacts = [
    {"name": "Sam", "phone": "555-0100", "city": "Lagos"},
    {"name": "Mina", "phone": "555-0142", "city": "Seoul"},
]
text = json.dumps(contacts)
print(text)
```

## Turn the text back into the list

`loads` turns JSON text back into a Python value. The call is `json.loads`.


```python
import json
contacts = [
    {"name": "Sam", "phone": "555-0100", "city": "Lagos"},
    {"name": "Mina", "phone": "555-0142", "city": "Seoul"},
    {"name": "Tina", "phone": "001-3938", "city": "Mumbai"},
]
text = json.dumps(contacts)
loaded = json.loads(text)
print(loaded[0]["name"])
print(len(loaded))
```

## Save the text in a file

On this PC, Jupyter's working folder is the `lessons` folder, the same folder as this notebook. A path next to the notebook is only the file name, such as `contacts.json`. A `lessons/` prefix makes `open()` raise `FileNotFoundError`.

`json.dumps` makes the text. Write that text into `contacts.json` with a `with` block.


```python
import json
contacts = [
    {"name": "Sam", "phone": "555-0100", "city": "Lagos"},
    {"name": "Mina", "phone": "555-0142", "city": "Seoul"},
]
text = json.dumps(contacts)
with open("contacts.json", "w", encoding="utf-8") as json_file:
    json_file.write(text)
with open("contacts.json", "r", encoding="utf-8") as json_file:
    print(json_file.read(), end="")
```

## Load the same list

Read `contacts.json`, then call `json.loads` on that text. You get the same contact list back.


```python
import json
with open("contacts.json", "r", encoding="utf-8") as json_file:
    text = json_file.read()
loaded = json.loads(text)
print(loaded[0]["name"])
print(len(loaded))
```
