# Plan, tool, observe, answer

An **agent** is a program that uses a tool before it answers. The four steps are plan, tool call, observe, and answer.

**Plan** chooses the tool name. A price question chooses `price`. Any other question chooses `stock`.

A **tool call** runs that tool. `price_of` returns a number. `in_stock` returns True or False.

**Observe** keeps the value the tool returned.

**Answer** writes the sentence from that value.

Run the cells from top to bottom. Each Try it is one step. These cells run in the notebook.


## Plan


```python
def price_of(name):
    prices = {"pen": 1.5, "notebook": 3.0, "pencil": 0.8}
    return prices[name]

def in_stock(name):
    stock = {"pen": True, "notebook": False, "pencil": True}
    return stock[name]

def plan(question):
    text = question.lower()
    if "cost" in text or "price" in text:
        return "price"
    return "stock"

def answer(name, tool_name, value):
    label = name.capitalize()
    if tool_name == "price":
        return "A " + label + " costs " + str(value) + "."
    if value:
        return "A " + label + " is in stock."
    return "A " + label + " is not in stock."
print(plan("What does a pen cost?"))
print(plan("Is a notebook in stock?"))
```

## Tool call


```python
def price_of(name):
    prices = {"pen": 1.5, "notebook": 3.0, "pencil": 0.8}
    return prices[name]

def in_stock(name):
    stock = {"pen": True, "notebook": False, "pencil": True}
    return stock[name]

def plan(question):
    text = question.lower()
    if "cost" in text or "price" in text:
        return "price"
    return "stock"

def answer(name, tool_name, value):
    label = name.capitalize()
    if tool_name == "price":
        return "A " + label + " costs " + str(value) + "."
    if value:
        return "A " + label + " is in stock."
    return "A " + label + " is not in stock."
print(price_of("pen"))
print(in_stock("notebook"))
```

## Observe and answer


```python
def price_of(name):
    prices = {"pen": 1.5, "notebook": 3.0, "pencil": 0.8}
    return prices[name]

def in_stock(name):
    stock = {"pen": True, "notebook": False, "pencil": True}
    return stock[name]

def plan(question):
    text = question.lower()
    if "cost" in text or "price" in text:
        return "price"
    return "stock"

def answer(name, tool_name, value):
    label = name.capitalize()
    if tool_name == "price":
        return "A " + label + " costs " + str(value) + "."
    if value:
        return "A " + label + " is in stock."
    return "A " + label + " is not in stock."
def run(question, name):
    tool_name = plan(question)
    if tool_name == "price":
        observed = price_of(name)
    else:
        observed = in_stock(name)
    return answer(name, tool_name, observed)

print(run("What does a pen cost?", "pen"))
print(run("Is a notebook in stock?", "notebook"))
```
