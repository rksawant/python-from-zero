# filter and map

Pen is in stock. Notebook is not. Each product is a dict with `name`, `price`, and `in_stock`.

**filter** keeps items when a function returns True. **map** builds a new value for each item. The function here is a lambda, the one-line function from lesson 31. `list` walks the result into a list you can print.

Run the cells from top to bottom. Each Try it is one step. These cells run in the notebook.


```python
products = [
    {"name": "Pen", "price": 1.5, "in_stock": True},
    {"name": "Notebook", "price": 3.0, "in_stock": False},
    {"name": "Ink", "price": 5.0, "in_stock": True},
]
kept = filter(lambda item: item["in_stock"], products)
for item in kept:
    print(item["name"])
```

### Try it

Replace `lambda item: True` with `lambda item: item["in_stock"]`. Before the change the cell prints `Pen` and `Notebook`. After the change it prints `Pen`.


```python
products = [
    {"name": "Pen", "price": 1.5, "in_stock": True},
    {"name": "Notebook", "price": 3.0, "in_stock": False},
    {"name": "Eraser", "price": 6.0, "in_stock": True},
    {"name": "Pencil", "price": 4.0, "in_stock": True},
]
res = filter(lambda item: item["price"] > 2 and item["in_stock"], products)
for item in res:
    print(item["name"])

res = filter(lambda x: "Pe" in x["name"], products)
rs = [print(item) for item in res]

"""
Basic syntax of map()
map(function, iterable)
function — what operation you want to perform.
iterable — a list, tuple, or another iterable.
map() — applies the function to every item.
list() — converts the resulting map iterator into a list.
"""

mapres = map(lambda x: x["name"] if "Pe" in x["name"] else "", products)
for item in mapres:
    print(item)

# Get only Product prices
mapres = map(lambda inp: inp["price"], products)
list(mapres)

# Display stock availability and its name
mapres = map(lambda inp: f"{inp["name"]} : Available" if inp["in_stock"] else f"{inp["name"]} : Not-Available", products)
list(mapres)
```

### Try it

Replace `lambda item: item` with `lambda item: item["name"]`. Before the change the cell prints the two dicts. After the change it prints `['Pen', 'Notebook']`.


```python
products = [
    {"name": "Pen", "price": 1.5, "in_stock": True},
    {"name": "Notebook", "price": 3.0, "in_stock": False},
]

#filter(): Used for filtering data
available = filter(lambda item: item["in_stock"], products)
print(f"filter(): {list(available)}")

names = map(lambda item: item, products)
print(list(names))

#Using map(): Used for calculations and conversions
names = list(map(lambda p: p["name"], products))
print(f"map(): {names}")

#Using list comprehension : In modern Python, list comprehensions are often easier to read
names = [p["name"] for p in products]
print(f"list comprehension: {names}")
```
