# filter and map

Pen is in stock. Notebook is not. Each product is a dict with `name`, `price`, and `in_stock`.

**filter** keeps items when a function returns True. **map** builds a new value for each item. The function here is a lambda, the one-line function from lesson 31. `list` walks the result into a list you can print.

Run the cells from top to bottom. Each Try it is one step. These cells run in the notebook.


```python
products = [
    {"name": "Pen", "price": 1.5, "in_stock": True},
    {"name": "Notebook", "price": 3.0, "in_stock": False},
]
kept = filter(lambda item: item["in_stock"], products)
for item in kept:
    print(item["name"])
```
