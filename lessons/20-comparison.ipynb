# Compare fields

Pen costs 1.5 and is in stock. Notebook costs 3.0 and is not in stock. Pencil costs 0.8 and is in stock. Three products make the comparisons easier to see.

A **filter** is the dict you pass to `find`. A **query operator** is a key that starts with `$`. It changes how one field is compared.

Each cell clears `products`, stores the three products, then prints the names that match. Run the cells from top to bottom. Each Try it is one step.


## $eq

`$eq` means equal. `{"name": {"$eq": "Pen"}}` keeps the product whose name is Pen.


### Try it

Replace the empty filter `{}` with `{"name": {"$eq": "Pen"}}`. Before the change the cell prints `Pen`, `Notebook`, and `Pencil`. After the change it prints `Pen`.


```python
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017", serverSelectionTimeoutMS=2000)
products = client["python_course"]["products"]
products.delete_many({})
products.insert_many([
    {"name": "Pen", "price": 1.5, "in_stock": True},
    {"name": "Notebook", "price": 3.0, "in_stock": False},
    {"name": "Pencil", "price": 0.8, "in_stock": True},
])
for doc in products.find({"name":{"$eq":"Pen"}}):
    print(doc["name"])
```
