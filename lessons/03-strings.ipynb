## Characters

A **character** is one piece of a string: a letter, a digit, a space, or a mark such as `!`. The space between `good` and `morning` is a character. It counts.


## How many characters

`len` is a **built-in** tool. Built-in means Python already has it, the way it already has `print` and `type`. Put a string inside the parentheses, and `len` gives the number of characters.


```python
message = "good morning"
print(len(message))
```

`good morning` has 12 characters: 4 in `good`, 1 space, and 7 in `morning`.

## Positions

Each character has a **position**. A position is also called an **index**. Counting starts at `0`, not at `1`. The first character is at position `0`. The second is at position `1`.

You ask for one character by writing the string's name, then **square brackets** `[ ]`, with the position inside. Square brackets are not parentheses.


```python
message = "good morning"
print(message[0])
print(message[5])
```

Position `0` is `g`. Position `5` is `m`, because the characters start `g`, `o`, `o`, `d`, space, `m`.

## A slice

A **slice** is a stretch of characters from one position up to another. You write two positions inside the square brackets, with a colon `:` between them. The colon here is part of the slice. It is not a punctuation mark for the sentence.


```python
message = "good morning ABC"
print(message[0:4])
```

The first number is the position where the slice starts. That character is included. The second number is the position where the slice stops. That character is not included. So `0:4` takes positions `0`, `1`, `2`, and `3`, which are the letters in `good`.


### Try it

## Uppercase and lowercase

A **method** is a tool that belongs to one value. You write the value, then a dot `.`, then the method name, then parentheses.

`.upper()` gives a new string in capital letters. `.lower()` gives a new string in small letters. The string you started with stays as it was, unless you store the new string in a name.


```python
message = "good morning"
print(message.upper())
print(message)
```

## Joining strings

`+` joins strings end to end. With numbers, `+` adds. With strings, `+` glues the text together. A space appears in the result only if one of the strings contains a space.


```python
first = "Ada"
last = "Lovelace"
print(first + " " + last)
```

## Put a value inside a sentence

An **f-string** is a string with the letter `f` immediately before the opening quote. Inside the string, **curly braces** `{ }` mark a spot where Python inserts a value. Curly braces are not parentheses and not square brackets.

The `f` has to touch the quote. `f"Hello"` is an f-string.


```python
name = "Sam"
print(f"Hello, {name}")
```
