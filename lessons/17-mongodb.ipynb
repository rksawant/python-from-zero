# Install MongoDB and open a connection

This lesson installs MongoDB on Windows, checks that the server is running, and opens a connection from Python.

Run the cells from top to bottom. Each Try it is one step.


## MongoDB

**MongoDB** is a program that stores data on this computer. **Community** is the free edition. **mongod** is the server program. It waits for other programs to connect.


## Install it with winget

**winget** is the Windows program that installs other programs.

Open PowerShell. The course folder is a fine place. Any other folder is fine too. This command does not use the private Python, so the line does not need to start with `(.venv)`.

```powershell
winget install --id MongoDB.Server -e
```

`winget` is the installer. `install` means add a program. `--id` means the next word is the package id. `MongoDB.Server` is the id of MongoDB Community. `-e` means that id must match exactly.

If winget asks you to agree, accept the agreements.


## What a running server looks like

A **service** is a program Windows keeps running in the background. The MongoDB service is named `MongoDB`.

Open PowerShell. Any folder works.

```powershell
Get-Service MongoDB
```

`Get-Service` asks Windows for one service. `MongoDB` is the service name.

`Status : Running` means mongod is running.

If Status is Stopped, start it:

```powershell
Start-Service MongoDB
```

`Start-Service` starts a service. `MongoDB` is the service name. Any folder works. If Windows refuses, close PowerShell, open it with Run as administrator, and run that command again.

A **port** is the number where a program waits. MongoDB's port is `27017`. **localhost** is this computer.

```powershell
Test-NetConnection -ComputerName localhost -Port 27017
```

`Test-NetConnection` checks an address. `-ComputerName` means the next word is the computer. `localhost` is this computer. `-Port` means the next number is the port. `27017` is MongoDB's port. Any folder works.

Read `TcpTestSucceeded`. True means mongod is accepting connections on port 27017. False means it is not.


## Install pymongo

**pymongo** is the library Python uses to talk to MongoDB. Its name is already in `requirements.txt`.

Open PowerShell. Move into the course folder, the folder that contains `requirements.txt`. Turn on the private Python so the line starts with `(.venv)`.

```powershell
pip install -r requirements.txt
```

`pip` is Python's installer. `install` means add the libraries. `-r` means read the package names from a file. `requirements.txt` is that file.


## Open a connection

**MongoClient** opens the connection. The address is `mongodb://localhost:27017`.

`MongoClient` is the call. The text in quotes is the address. `mongodb://` marks the address as MongoDB. `localhost` is this computer. `27017` is the port.

`serverSelectionTimeoutMS=2000` tells MongoClient to wait 2 seconds, then stop.

**ping** asks the server to answer. `client.admin.command("ping")` sends that ask. A running server answers with `ok` equal to `1.0`. The cell prints `MongoDB is up`.

If mongod is stopped, pymongo raises **ServerSelectionTimeoutError**. On Windows the message includes `[WinError 10061]` and says the connection was refused. Start the MongoDB service, then run the cell again.


````python
from pymongo import MongoClient
from pymongo.errors import ServerSelectionTimeoutError

client = MongoClient("mongodb://localhost:27017", serverSelectionTimeoutMS=2000)
try:
    client.admin.command("ping")
    print("MongoDB is up")
except ServerSelectionTimeoutError as error:
    print(type(error).__name__)
    detail = str(error)
    if "actively refused it" in detail:
        print("localhost:27017: [WinError 10061] No connection could be made because the target machine actively refused it")
    else:
        print("MongoDB is not running at localhost:27017")
````
