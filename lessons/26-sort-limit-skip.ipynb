# Sort, limit, and skip

Pen costs 1.5, Notebook costs 3.0, and Pencil costs 0.8. `find` returns a cursor. **sort**, **limit**, and **skip** are calls on that cursor.

`sort("price", 1)` orders by price. `1` means the smallest price first. `-1` means the largest price first.

`limit(2)` keeps the first 2 documents after the sort. `skip(1)` drops the first document after the sort.

You can chain them: `find({}).sort("price", 1).limit(2)`.

Run the cells from top to bottom. Each Try it is one step.


### Try it

Replace `.sort("name", 1)` with `.sort("price", 1)`. Before the change the names are alphabetical: `Notebook`, `Pen`, `Pencil`. After the change the smallest price is first: `Pencil`, `Pen`, `Notebook`.


```python
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017", serverSelectionTimeoutMS=2000)
products = client["python_course"]["products"]
products.delete_many({})
products.insert_many([
    {"name": "Pen", "price": 1.5, "in_stock": True},
    {"name": "Notebook", "price": 3.0, "in_stock": False},
    {"name": "Pencil", "price": 0.8, "in_stock": True},
])
for doc in products.find({}).sort("name", -1):
    print(doc["name"])

for item in products.find({}).sort("price", 1):
    print(item["price"])
```
