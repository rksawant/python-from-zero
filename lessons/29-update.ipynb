# Update products

**update_one** changes the first document that matches the filter. **update_many** changes every document that matches.

**$set** sets field values and leaves the other fields in place. The call is `update_one(filter, {"$set": {"price": 2.0}})`.

A filter can use more than one condition. `{"name": "Pen", "in_stock": True}` matches only a Pen that is in stock.

Each cell clears `products` and stores Pen, Notebook, and Pencil again. Run the cells from top to bottom. Each Try it is one step.


```python
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017", serverSelectionTimeoutMS=2000)
products = client["python_course"]["products"]
products.delete_many({})
products.insert_many([
    {"name": "Pen", "price": 1.5, "in_stock": True},
    {"name": "Notebook", "price": 3.0, "in_stock": False},
    {"name": "Pencil", "price": 0.8, "in_stock": True},
])
products.update_one(
    {"name": "Pen", "in_stock": True},
    {"$set": {"price": 2.0}},
)
doc = products.find_one({"name": "Pen"})
print(doc["name"])
print(doc["price"])
print(doc["in_stock"])

products.update_one(
    {"price":{"$lte":2}},
    {"$set":{"price":5}}
)

print(products.find_one({"price":{"$gt":1}}))
```

## update_many

`update_many({"in_stock": True}, {"$set": {"shelf": "A"}})` sets `shelf` to `A` on every in-stock product. Pen and Pencil are in stock. Notebook is not, so it stays without a shelf.


```python
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017", serverSelectionTimeoutMS=2000)
products = client["python_course"]["products"]
products.delete_many({})
products.insert_many([
    {"name": "Pen", "price": 1.5, "in_stock": True},
    {"name": "Notebook", "price": 3.0, "in_stock": False},
    {"name": "Pencil", "price": 0.8, "in_stock": True},
])
products.update_many(
    {"in_stock": True},
    {"$set": {"shelf": "A"}},
)
print(products.count_documents({"shelf": "A"}))
```

Replace the filter `{"in_stock": True}` with `{"in_stock": True, "price": {"$lt": 1}}`. Both conditions have to match, so only Pencil changes. Before the change the cell prints `Pencil A` and `Pen A`. After the change it prints `Pencil B` and `Pen None`.


```python
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017", serverSelectionTimeoutMS=2000)
products = client["python_course"]["products"]
products.delete_many({})
products.insert_many([
    {"name": "Pen", "price": 1.5, "in_stock": True},
    {"name": "Notebook", "price": 3.0, "in_stock": False},
    {"name": "Pencil", "price": 0.8, "in_stock": True},
])
products.update_many(
    {"in_stock": True},
    {"$set": {"shelf": "B"}},
)
for name in ("Pencil", "Pen"):
    doc = products.find_one({"name": name})
    print(name, doc.get("shelf"))
```

```python
# Solution
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017", serverSelectionTimeoutMS=2000)
products = client["python_course"]["products"]
products.delete_many({})
products.insert_many([
    {"name": "Pen", "price": 1.5, "in_stock": True},
    {"name": "Notebook", "price": 3.0, "in_stock": False},
    {"name": "Pencil", "price": 0.8, "in_stock": True},
])
products.update_many(
    {"in_stock": True, "price": {"$lt": 1}},
    {"$set": {"shelf": "B"}},
)
for name in ("Pencil", "Pen"):
    doc = products.find_one({"name": name})
    print(name, doc.get("shelf"))
```
