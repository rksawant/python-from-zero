# Dict functions

A **dict** is a dictionary. Lesson 4 used the full word. This lesson uses the short name. A **key** is the label. A **value** is the item stored under that label.

The example is one product: name Pen, price 1.5, in stock True.

Run the cells from top to bottom. Each Try it is one step. These cells run in the notebook.


```python
product = {"name": "Pen", "price": 1.5, "in_stock": True}
print(product["name"])
print(product["price"])
print(product["in_stock"])
```

## keys

`keys` returns the labels. `list` puts them on one line you can print.


### Try it

Replace `labels = None` with `labels = list(product.keys())`. Before the change the cell prints `None`. After the change it prints `['name', 'price', 'in_stock']`.


```python
product = {"name": "Pen", "price": 1.5, "in_stock": True}
labels = None
print(labels)
```

### Solution

Try the exercise above first. Unfold the next cell the same way as in lesson 1: click the blue bar on its left, then press Shift+Enter.


```python
# Solution
product = {"name": "Pen", "price": 1.5, "in_stock": True}
labels = list(product.keys())
print(labels)
```

## items

`items` returns each key together with its value. The `for` line names the key and the value, one pair at a time.


### Try it

Replace `pass` with the two lines `for key, value in product.items():` and `print(key, value)`. Before the change the cell prints nothing. After the change it prints the three pairs.


```python
product = {"name": "Pen", "price": 1.5, "in_stock": True}
pass
```

### Solution

Try the exercise above first. Unfold the next cell the same way as in lesson 1: click the blue bar on its left, then press Shift+Enter.


```python
# Solution
product = {"name": "Pen", "price": 1.5, "in_stock": True}
for key, value in product.items():
    print(key, value)
```

## get

`get` returns the value for a key. The second input is what you get when the key is missing. `get` does not add the key.


### Try it

Replace `answer = None` with `answer = product.get("city", "Lagos")`. Before the change the cell prints `None` and `False`. After the change it prints `Lagos` and `False`.


```python
product = {"name": "Pen", "price": 1.5, "in_stock": True}
answer = None
print(answer)
print("city" in product)
```

### Solution

Try the exercise above first. Unfold the next cell the same way as in lesson 1: click the blue bar on its left, then press Shift+Enter.


```python
# Solution
product = {"name": "Pen", "price": 1.5, "in_stock": True}
answer = product.get("city", "Lagos")
print(answer)
print("city" in product)
```

## update

`update` copies pairs into the dict. A key that is already there gets the new value. A new key is added.


### Try it

Add the line `product.update({"price": 2.0, "city": "Lagos"})` above the first print. Before the change the cell prints `1.5` and `None`. After the change it prints `2.0` and `Lagos`.


```python
product = {"name": "Pen", "price": 1.5, "in_stock": True}
print(product["price"])
print(product.get("city"))
```

### Solution

Try the exercise above first. Unfold the next cell the same way as in lesson 1: click the blue bar on its left, then press Shift+Enter.


```python
# Solution
product = {"name": "Pen", "price": 1.5, "in_stock": True}
product.update({"price": 2.0, "city": "Lagos"})
print(product["price"])
print(product.get("city"))
```

## pop

`pop` removes one key and returns the value that was stored there.


### Try it

Replace `price = None` with `price = product.pop("price")`. Before the change the cell prints `None` and `True`. After the change it prints `1.5` and `False`.


```python
product = {"name": "Pen", "price": 1.5, "in_stock": True}
price = None
print(price)
print("price" in product)
```

### Solution

Try the exercise above first. Unfold the next cell the same way as in lesson 1: click the blue bar on its left, then press Shift+Enter.


```python
# Solution
product = {"name": "Pen", "price": 1.5, "in_stock": True}
price = product.pop("price")
print(price)
print("price" in product)
```

## popitem

`popitem` removes one pair and returns it. The pair is the last one that was added. In this dict, that pair is the price.


### Try it

Replace `pair = None` with `pair = product.popitem()`. Before the change the cell prints `None` and the whole dict. After the change it prints `('price', 1.5)` and `{'name': 'Pen'}`.


```python
product = {"name": "Pen", "price": 1.5}
pair = None
print(pair)
print(product)
```

### Solution

Try the exercise above first. Unfold the next cell the same way as in lesson 1: click the blue bar on its left, then press Shift+Enter.


```python
# Solution
product = {"name": "Pen", "price": 1.5}
pair = product.popitem()
print(pair)
print(product)
```

## copy

`copy` returns a new dict with the same pairs. Changing the new dict leaves the original price at 1.5.


### Try it

Replace `other = product` with `other = product.copy()`. Before the change the cell prints `2.0`, because both names point at one dict. After the change it prints `1.5`.


```python
product = {"name": "Pen", "price": 1.5, "in_stock": True}
other = product
other["price"] = 2.0
print(product["price"])
```

### Solution

Try the exercise above first. Unfold the next cell the same way as in lesson 1: click the blue bar on its left, then press Shift+Enter.


```python
# Solution
product = {"name": "Pen", "price": 1.5, "in_stock": True}
other = product.copy()
other["price"] = 2.0
print(product["price"])
```

## fromkeys

`dict.fromkeys` builds a new dict. The first input is the keys. The second input is the value stored for every one of those keys. `dict` in front of the dot means you call it on the dict type.


### Try it

Replace `fresh = None` with `fresh = dict.fromkeys(["name", "city"], "unknown")`. Before the change the cell prints `None`. After the change it prints `{'name': 'unknown', 'city': 'unknown'}`.


```python
fresh = None
print(fresh)
```

### Solution

Try the exercise above first. Unfold the next cell the same way as in lesson 1: click the blue bar on its left, then press Shift+Enter.


```python
# Solution
fresh = dict.fromkeys(["name", "city"], "unknown")
print(fresh)
```
