The products are still Pen at 1.5 in stock, Notebook at 3.0 not in stock, and Pencil at 0.8 in stock.

A **logical operator** combines conditions. `$and`, `$or`, `$nor`, and `$not` are the four in this lesson. Two keys in one filter are combined as well, with no operator written on the outside.

## Two conditions in one filter

Two keys in one filter both have to match. `{"price": {"$gt": 1}, "in_stock": True}` keeps a product only when the price is greater than 1 and `in_stock` is True. That product is Pen.


### Try it

Replace `{"in_stock": True}` with `{"price": {"$gt": 1}, "in_stock": True}`. Before the change the cell prints `Pen` and `Pencil`. After the change it prints `Pen`.


```python
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017", serverSelectionTimeoutMS=2000)
products = client["python_course"]["products"]
products.delete_many({})
products.insert_many([
    {"name": "Pen", "price": 1.5, "in_stock": True},
    {"name": "Notebook", "price": 3.0, "in_stock": False},
    {"name": "Pencil", "price": 0.8, "in_stock": True},
])
for doc in products.find({"in_stock": True}):
    print(doc["name"])
```

## $or

`$or` takes a list of filters. One filter in the list has to match. `{"$or": [{"price": {"$gt": 2}}, {"name": "Pencil"}]}` keeps Notebook, because its price is greater than 2, and Pencil, because its name is Pencil.


### Try it

Replace `{"name": "Pencil"}` with `{"$or": [{"price": {"$gt": 2}}, {"name": "Pencil"}]}`. Before the change the cell prints `Pencil`. After the change it prints `Notebook` and `Pencil`.


```python
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017", serverSelectionTimeoutMS=2000)
products = client["python_course"]["products"]
products.delete_many({})
products.insert_many([
    {"name": "Pen", "price": 1.5, "in_stock": True},
    {"name": "Notebook", "price": 3.0, "in_stock": False},
    {"name": "Pencil", "price": 0.8, "in_stock": True},
])
for doc in products.find({"name": "Pencil"}):
    print(doc["name"])
```

`$nor` takes a list of filters. A document matches when it matches none of those filters. `{"$nor": [{"name": "Pen"}, {"in_stock": False}]}` leaves out Pen, and leaves out every product that is not in stock. Pencil remains.

Replace the empty filter `{}` with `{"$nor": [{"name": "Pen"}, {"in_stock": False}]}`. Before the change the cell prints `Pen`, `Notebook`, and `Pencil`. After the change it prints `Pencil`.

## $not

`$not` wraps one comparison. The field matches when that comparison fails. `{"price": {"$not": {"$gt": 1}}}` keeps prices that are not greater than 1. That price is Pencil at 0.8.


### Try it

Replace `{"price": {"$gt": 1}}` with `{"price": {"$not": {"$gt": 1}}}`. Before the change the cell prints `Pen` and `Notebook`. After the change it prints `Pencil`.


```python
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017", serverSelectionTimeoutMS=2000)
products = client["python_course"]["products"]
products.delete_many({})
products.insert_many([
    {"name": "Pen", "price": 1.5, "in_stock": True},
    {"name": "Notebook", "price": 3.0, "in_stock": False},
    {"name": "Pencil", "price": 0.8, "in_stock": True},
])
for doc in products.find({"price": {"$gt": 1}}):
    print(doc["name"])
```

```python
# Solution
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017", serverSelectionTimeoutMS=2000)
products = client["python_course"]["products"]
products.delete_many({})
products.insert_many([
    {"name": "Pen", "price": 1.5, "in_stock": True},
    {"name": "Notebook", "price": 3.0, "in_stock": False},
    {"name": "Pencil", "price": 0.8, "in_stock": True},
])
for doc in products.find({"price": {"$not": {"$gt": 1}}}):
    print(doc["name"])
```
