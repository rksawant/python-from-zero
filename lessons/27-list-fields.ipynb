# A field that holds a list

A product document can hold a list. `tags` is a list of words. Pen has `ink` and `desk`. Notebook has `paper` and `desk`. Pencil has `wood`.

`stores` is a list of documents. Each one has `city` and `qty`. Pen is in Lagos with qty 2 and in Seoul with qty 0. Notebook is in Lagos with qty 1. Pencil is in Accra with qty 5.

Run the cells from top to bottom. Each Try it is one step.


```python
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017", serverSelectionTimeoutMS=2000)
products = client["python_course"]["products"]
products.delete_many({})
products.insert_many([
    {
        "name": "Pen",
        "price": 1.5,
        "in_stock": True,
        "tags": ["ink", "desk"],
        "stores": [
            {"city": "Lagos", "qty": 2},
            {"city": "Seoul", "qty": 0},
        ],
    },
    {
        "name": "Notebook",
        "price": 3.0,
        "in_stock": False,
        "tags": ["paper", "desk"],
        "stores": [{"city": "Lagos", "qty": 1}],
    },
    {
        "name": "Pencil",
        "price": 0.8,
        "in_stock": True,
        "tags": ["wood"],
        "stores": [{"city": "Accra", "qty": 5}],
    },
])
doc = products.find_one({"name": "Pen"})
print(doc["tags"])
print(doc["stores"])
```

## $size

## $regex

A **regex** is a text pattern. `$regex` matches a field against that pattern. `^` means the start of the text. `^P` matches a name that starts with P.


### Try it

## $elemMatch

`$elemMatch` matches when one item in the list meets every condition inside it. Pen has one store that is Lagos and has qty greater than 1. Notebook's Lagos store has qty 1, so that item fails qty greater than 1.


### Try it

Replace `{"stores": {"$elemMatch": {"city": "Lagos", "qty": {"$gt": 0}}}}` with `{"stores": {"$elemMatch": {"city": "Lagos", "qty": {"$gt": 1}}}}`. Before the change the cell prints `Pen` and `Notebook`. After the change it prints `Pen`.


```python
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017", serverSelectionTimeoutMS=2000)
products = client["python_course"]["products"]
products.delete_many({})
products.insert_many([
    {
        "name": "Pen",
        "price": 1.5,
        "in_stock": True,
        "tags": ["ink", "desk"],
        "stores": [
            {"city": "Lagos", "qty": 2},
            {"city": "Seoul", "qty": 0},
        ],
    },
    {
        "name": "Notebook",
        "price": 3.0,
        "in_stock": False,
        "tags": ["paper", "desk"],
        "stores": [{"city": "Lagos", "qty": 1}],
    },
    {
        "name": "Pencil",
        "price": 0.8,
        "in_stock": True,
        "tags": ["wood"],
        "stores": [{"city": "Accra", "qty": 5}],
    },
])
for doc in products.find({"stores": {"$elemMatch": {"city": "Lagos", "qty": {"$gt": 0}}}}):
    print(doc["name"])
```
