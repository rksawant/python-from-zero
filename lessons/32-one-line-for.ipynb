# A one-line for with an if

The names are Pen, Notebook, and Pencil. Pen has 3 letters. The other names are longer.

A **one-line for with an if** builds a new list in one line.

`short = [name for name in names if len(name) <= 3]`

`name` at the start is the value that goes into the new list. `for name in names` walks each name. `if len(name) <= 3` keeps a name only when it has 3 letters or fewer.

Run the cells from top to bottom. Each Try it is one step. These cells run in the notebook.


```python
names = ["Pen", "Notebook", "Pencil"]
short = [name for name in names if len(name) <= 3]
print(short)
```
