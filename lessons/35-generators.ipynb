# Generators

A **generator** is a function that pauses. **yield** hands back one value and waits. The next ask continues after that `yield`.

Run the cells from top to bottom. Each Try it is one step. These cells run in the notebook.


```python
def prices():
    yield 1.5
    yield 3.0

for price in prices():
    print(price)
```

### Try it

Replace `pass` with `yield 0.8`. Before the change the cell prints `1.5` and `3.0`. After the change it also prints `0.8`.


```python
def prices():
    yield 1.5
    yield 3.0
    pass

for price in prices():
    print(price)
```

### Solution

Try the exercise above first. Unfold the next cell the same way as in lesson 1: click the blue bar on its left, then press Shift+Enter.


```python
# Solution
def prices():
    yield 1.5
    yield 3.0
    yield 0.8

for price in prices():
    print(price)
```

## next

**next** asks a generator for one more value. `made = prices()` starts the generator. `next(made)` returns the next yielded price.


### Try it

Add a second line `print(next(made))` under the first print. Before the change the cell prints `1.5`. After the change it prints `1.5` and `3.0`.


```python
def prices():
    yield 1.5
    yield 3.0

made = prices()
print(next(made))
```

### Solution

Try the exercise above first. Unfold the next cell the same way as in lesson 1: click the blue bar on its left, then press Shift+Enter.


```python
# Solution
def prices():
    yield 1.5
    yield 3.0

made = prices()
print(next(made))
print(next(made))
```
