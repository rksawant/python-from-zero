# A small LangGraph agent

**LangGraph** keeps a product question moving through steps. The **state** is the dict those steps carry. This agent stores `question`, `route`, and `answer`.

A **node** is one function. It reads the state and returns the next state.

A **tool** is a function a node calls when it needs a fact. `price_of` returns the price of a product name.

An **edge** is the line from one node to the next. A **conditional edge** picks the next node from the state. A cost question goes to the price node. Any other question goes to the stock node.

A **graph** is the nodes and edges together. `StateGraph` builds it. `compile` makes it runnable. `invoke` runs it once.

Run the cells from top to bottom. Each Try it is one step. These cells run in the notebook.


## The state


```python
state = {"question": "What does a pen cost?", "route": "", "answer": ""}
print(state["question"])
print(state["route"])
print(state["answer"])
```

## The price tool


```python
def price_of(name):
    prices = {"pen": 1.5, "notebook": 3.0, "pencil": 0.8}
    return prices[name]

print(price_of("pen"))
print(price_of("notebook"))
print(price_of("pencil"))
```

## The graph


```python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END

class State(TypedDict):
    question: str
    route: str
    answer: str

def price_of(name):
    prices = {"pen": 1.5, "notebook": 3.0, "pencil": 0.8}
    return prices[name]

def route_question(state: State) -> State:
    text = state["question"].lower()
    if "cost" in text or "price" in text:
        route = "price"
    else:
        route = "stock"
    return {"question": state["question"], "route": route, "answer": state["answer"]}

def price_node(state: State) -> State:
    return {"question": state["question"], "route": state["route"], "answer": "A Pen costs " + str(price_of("pen")) + "."}

def stock_node(state: State) -> State:
    return {"question": state["question"], "route": state["route"], "answer": "A Notebook is not in stock."}

def choose(state: State) -> str:
    return state["route"]

graph = StateGraph(State)
graph.add_node("route", route_question)
graph.add_node("price", price_node)
graph.add_node("stock", stock_node)
graph.add_edge(START, "route")
graph.add_conditional_edges("route", choose, {"price": "price", "stock": "stock"})
graph.add_edge("price", END)
graph.add_edge("stock", END)
app = graph.compile()
print(app.invoke({"question": "What does a pen cost?", "route": "", "answer": ""})["answer"])
print(app.invoke({"question": "Is a notebook in stock?", "route": "", "answer": ""})["answer"])
```
