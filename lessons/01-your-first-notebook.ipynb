# Your first notebook

This lesson is about the page you are looking at. You will learn what Jupyter is, what a cell is, how to run a cell, and how to read the output. The first Python instruction comes only after those ideas, in the section **Check that the kernel works**.


## What Jupyter is

**Jupyter** is a program for reading lessons like this one and running small instructions one box at a time.

**JupyterLab** is the window Jupyter opens. You will usually see a list of files on the left and this lesson on the right.

**Python** is the language the later lessons teach. A language, here, is a set of instructions a computer can follow. You do not need to write Python to understand this page. Keep reading from top to bottom.


## What a notebook is

This whole page is a **notebook**. A notebook is a document made of boxes stacked from top to bottom. You read the boxes in order. Some boxes explain an idea. Some boxes hold instructions you can run.

This notebook is a **file**, a saved document on the computer. Its name is `01-your-first-notebook.ipynb`. The ending `.ipynb` means "this file is a notebook."


## What a cell is

Each box is a **cell**.

A **text cell** holds sentences, headings, and bullet lists, like the words you are reading now. You read a text cell. Running it is optional. If you double-click a text cell, extra formatting marks appear, such as `#` at the start of a heading. Those marks shape this page. Press Shift+Enter while that cell is selected to return to the reading view. The next section shows what Shift+Enter does.

A **code cell** holds instructions for the computer. You can recognize it by the label on its left: `In [ ]:`. The brackets start empty. Jupyter fills them after you run the cell. You do not type the `In [ ]:` label yourself. Jupyter puts it there.


## How to run a cell

To **run** a cell means to ask the computer to follow the instructions in that cell right now.

1. Click inside the code cell. A blue bar appears on its left edge. That bar means the cell is selected.
2. Press **Shift+Enter**. Hold the Shift key, and while you are holding it, press Enter.

Jupyter then runs the cell and moves the selection to the next cell. That move is expected.

You can also run the selected cell with the toolbar. The **toolbar** is the row of buttons above the notebook. Move the pointer over the buttons until a small label appears. The run button is a triangle, and its label says **Run the selected cells and advance**.

The first code cell is in **Check that the kernel works**, a few sections below. Read the next two sections, then come back to that cell and run it.


## How to read the output

**Output** is what the computer shows directly under a code cell after you run it. It is the reply to that cell.

Read a code cell in this order:

- Look at the left label. `In [ ]:` with empty brackets means this cell has not been run during this visit.
- After a run, the label looks like `In [1]:`. The number counts code cells the computer has finished. The next cell you run gets the next number, such as `In [2]:`.
- Then read the text under the cell. That text is the output.

If you run the same cell again, the number goes up, and the new output replaces the old output.


## What the kernel is

The **kernel** is the part of Jupyter that runs code cells. It stays on while you work. It also remembers the results of cells you have already run. Later lessons use that memory.

When a code cell runs and output appears under it, the kernel is working.

Sometimes the label shows an asterisk, like `In [*]:`. That means the kernel is still busy with that cell. Wait a moment. If it stays that way, open the **Kernel** menu at the top of JupyterLab and choose **Interrupt Kernel**. That asks the kernel to stop the busy cell.

If the cell is still stuck, choose **Restart Kernel**. A window asks you to confirm and says that remembered results will be lost. Click **Restart**. Then run the notebook again from the top. You should not need that menu in this lesson.


## Save the notebook

The notebook keeps what you type after you save it. Press **Ctrl+S**. On a Mac, press **Command+S**. Jupyter also saves the notebook every so often on its own. Ctrl+S is the sure way when you have just changed a cell.

You can also open the **File** menu and choose the item that starts with **Save**.
