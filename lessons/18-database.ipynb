## Collection

A **collection** is a named group of documents inside a database. The name in this course is `products`.

A **document** is one product. It has `name`, `price`, and `in_stock`, the same fields as Product in lesson 16.

`database["products"]` chooses that collection. `create_collection` creates the empty collection on the server now. The `if` line skips that call when `products` is already there, so the cell can run again.


### Try it

Replace `products = None` with `products = database["products"]`. Before the change the cell prints `None`. After the change it prints `products`.


```python
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017", serverSelectionTimeoutMS=2000)
database = client["python_course"]
products = database["products"]
print(None if products is None else products.name)
```

### Solution

Try the exercise above first. Unfold the next cell the same way as in lesson 1: click the blue bar on its left, then press Shift+Enter.


```python
# Solution
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017", serverSelectionTimeoutMS=2000)
database = client["python_course"]
products = database["products"]
print(None if products is None else products.name)
```

### Try it

Replace `pass` with `database.create_collection("products")`. The first time, before the change, the cell prints `False`. After the change it prints `True`. A later run still prints `True`, because the collection is already there.


```python
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017", serverSelectionTimeoutMS=2000)
database = client["python_course"]
if "products" not in database.list_collection_names():
    pass
print("products" in database.list_collection_names())
```

### Solution

Try the exercise above first. Unfold the next cell the same way as in lesson 1: click the blue bar on its left, then press Shift+Enter.


```python
# Solution
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017", serverSelectionTimeoutMS=2000)
database = client["python_course"]
if "products" not in database.list_collection_names():
    database.create_collection("products")
print("products" in database.list_collection_names())
```
