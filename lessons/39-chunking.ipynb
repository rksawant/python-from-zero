## Fixed

**Fixed** chunks take the same number of characters each time. `size` is that number. The next chunk starts where the previous one ends.


### Try it

```python
document = """Pen costs 1.5 and is in stock. Pen is on the front desk. The shop is closed on Sunday.

Notebook costs 3.0 and is not in stock.
"""
def fixed_chunks(text, size):
    chunks = []
    start = 0
    while start < len(text):
        chunks.append(text[start:start + size])
        start += size
    return chunks
print(len(fixed_chunks(document, 40)))
```

## Overlap

**Overlap** means the next chunk starts before the previous chunk ends, so a few characters are in both. `overlap` is how many characters are shared. `step` is `size` minus `overlap`.


### Try it

Replace `0` with `15`. The size stays `40`. Before the change the cell prints `4`. After the change it prints `5`.


```python
document = """Pen costs 1.5 and is in stock. Pen is on the front desk. The shop is closed on Sunday.

Notebook costs 3.0 and is not in stock.
"""
def overlap_chunks(text, size, overlap):
    chunks = []
    start = 0
    step = size - overlap
    while start < len(text):
        chunks.append(text[start:start + size])
        if start + size >= len(text):
            break
        start += step
    return chunks
print(len(overlap_chunks(document, 50, 0)))
```

## Paragraph

**Paragraph** chunks split on a blank line. A blank line is two line breaks in a row.


### Try it

Replace `chunks = []` with `chunks = paragraph_chunks(document)`. Before the change the cell prints `0`. After the change it prints `2`.


```python
document = """Pen costs 1.5 and is in stock. Pen is on the front desk. The shop is closed on Sunday.

Notebook costs 3.0 and is not in stock.
"""
def paragraph_chunks(text):
    chunks = []
    for part in text.split("\n\n"):
        part = part.strip()
        if part:
            chunks.append(part)
    return chunks
chunks = paragraph_chunks(document)
print(len(chunks))
```

## Semantic

**Semantic** chunks split where the topic changes. This lesson treats a topic as a name in the list. Two sentences stay in one chunk when they share a name. The shop sentence shares neither Pen nor Notebook, so it starts its own chunk.


### Try it

Replace `names = []` with `names = ["Pen", "Notebook"]`. Before the change every sentence is its own chunk, so the cell prints `4`. After the change the two Pen sentences stay together, so the cell prints `3`.


```python
document = """Pen costs 1.5 and is in stock. Pen is on the front desk. The shop is closed on Sunday.

Notebook costs 3.0 and is not in stock.
"""
def sentence_chunks(text):
    flat = " ".join(text.split())
    parts = flat.split(". ")
    chunks = []
    for part in parts:
        part = part.strip()
        if not part:
            continue
        if not part.endswith("."):
            part += "."
        chunks.append(part)
    return chunks
def semantic_chunks(text, names):
    sentences = sentence_chunks(text)
    chunks = []
    current = []
    for sentence in sentences:
        if not current:
            current.append(sentence)
            continue
        previous = " ".join(current)
        shares = False
        for name in names:
            if name in previous and name in sentence:
                shares = True
        if shares:
            current.append(sentence)
        else:
            chunks.append(" ".join(current))
            current = [sentence]
    if current:
        chunks.append(" ".join(current))
    return chunks
names = ["Pen"]
print(len(semantic_chunks(document, names)))
```
