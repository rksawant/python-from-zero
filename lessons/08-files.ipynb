## A path

A **text file** is a file that holds text.

A **path** is the directions to a file. A **relative path** starts from the course folder, the folder where you started Jupyter. It does not start with a drive letter. `lessons/note.txt` means the file `note.txt` inside the `lessons` folder, next to this notebook. The slash `/` works on Windows, Mac, and Linux.


## Write a file and close it

`open` asks Python to use a file. The second value is the **mode**. `"w"` means write. Write mode creates the file when it is missing. When the file is already there, write mode replaces the old text.

`with` opens the file and closes it when the indented lines finish. **Closing** means Python finishes saving and lets go of the file. You do not add a separate close line. The name after `as` is the open file for those indented lines.

`encoding="utf-8"` tells Python how to store letters. Use it every time you open a text file in this course.

`write` is a method on the open file. It takes one string. Inside a string, `\n` is a **newline**. It starts a new line in the file.


```python
with open("note.txt", "w", encoding="utf-8") as note:
    note.write("Hello from the file.\n")
print("saved")
```

After this cell runs, `note.txt` is in the `lessons` folder, next to this notebook. The `with` line already closed it.

## Read a file

Mode `"r"` means read. `read` is a method on the open file. It returns the whole file as one string.

`end=""` is a setting for `print`. It tells `print` not to add a second newline, because the file already ends with one.


```python
with open("note.txt", "r", encoding="utf-8") as note:
    print(note.read(), end="")
```

## Replace the old text

Opening the same path again with mode `"w"` replaces whatever was in the file. One `write` can hold several lines. Each `\n` starts the next line.


```python
with open("note.txt", "w", encoding="utf-8") as note:
    note.write("tea\nwater\n")
with open("note.txt", "r", encoding="utf-8") as note:
    print(note.read(), end="")
```
