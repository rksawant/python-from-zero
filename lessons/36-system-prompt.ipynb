# A system prompt

A **system prompt** is the instruction text you give a model before the user's question. It holds the rules for the answer.

The helper in this lesson answers questions about products. It may use only `name`, `price`, and `in_stock`. It answers in one sentence. If the product is not in the list, it says it does not have it.

Run the cells from top to bottom. Each Try it is one step. These cells run in the notebook.


```python
rules = """You answer questions about products.
Use only name, price, and in_stock.
Answer in one sentence.
If the product is not in the list, say you do not have it.
"""
print(rules)
```

## Zero-shot

**Zero-shot** means the prompt has the rules and one question, and no sample answer.


### Try it

Replace `prompt = rules` with `prompt = rules + "\nQuestion: " + question`. Before the change the cell does not print `Question: How much is a Pen?`. After the change it does.


```python
rules = """You answer questions about products.
Use only name, price, and in_stock.
Answer in one sentence.
If the product is not in the list, say you do not have it.
"""
question = "How much is a Pen?"
prompt = rules
print(prompt)

prompt = rules + "\nQuestion: " + question
print(prompt)
```

### Solution

Try the exercise above first. Unfold the next cell the same way as in lesson 1: click the blue bar on its left, then press Shift+Enter.


```python
# Solution
rules = """You answer questions about products.
Use only name, price, and in_stock.
Answer in one sentence.
If the product is not in the list, say you do not have it.
"""
question = "How much is a Pen?"
prompt = rules + "\nQuestion: " + question
print(prompt)
```

## Few-shot

**Few-shot** means the prompt also includes a few sample questions and answers before the real question.


### Try it

Replace `prompt = rules + "\nQuestion: " + question` with `prompt = rules + samples + "\nQuestion: " + question`. Before the change the cell has no sample answer. After the change it prints `Answer: A Notebook costs 3.0.`


```python
rules = """You answer questions about products.
Use only name, price, and in_stock.
Answer in one sentence.
If the product is not in the list, say you do not have it.
"""
samples = """
Question: How much is a Notebook?
Answer: A Notebook costs 3.0.

Question: Is a Pencil in stock?
Answer: Yes. A Pencil is in stock.
"""
question = "How much is a Pen?"
prompt = rules + "\nQuestion: " + question
print(prompt)

prompt = prompt + samples + "\nQuestion: " + question
print(prompt)
```

### Solution

Try the exercise above first. Unfold the next cell the same way as in lesson 1: click the blue bar on its left, then press Shift+Enter.


```python
# Solution
rules = """You answer questions about products.
Use only name, price, and in_stock.
Answer in one sentence.
If the product is not in the list, say you do not have it.
"""
samples = """
Question: How much is a Notebook?
Answer: A Notebook costs 3.0.

Question: Is a Pencil in stock?
Answer: Yes. A Pencil is in stock.
"""
question = "How much is a Pen?"
prompt = rules + samples + "\nQuestion: " + question
print(prompt)
```
